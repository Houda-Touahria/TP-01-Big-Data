```python
url_quotes = "https://todayinsci.com/Quotations/QuotationsIndex.htm"

response_quotes = requests.get(url_quotes)

print(response_quotes.status_code)
```

نستخرج كل اقتباسات انشتاين

```python
quotes_einstein = soup_einstein.find_all("div", class_="quotation")

print("عدد الاقتباسات التي وجدناها:", len(quotes_einstein))
```

نعرض اول 5 اقتباسات

```python
for i, quote in enumerate(quotes_einstein[:5], start=1):
    print(f"\nQuote {i}:")
    print(quote.get_text(" ", strip=True))
```

```python
from urllib.parse import urljoin

scientists = []

for link in links_quotes:
    name = link.get_text(" ", strip=True)
    href = link.get("href")

    if href and href.endswith("-Quotations.htm"):
        full_url = urljoin(url_quotes, href)

        scientists.append({
            "name": name,
            "url": full_url
        })

print("عدد العلماء:", len(scientists))

for scientist in scientists[:10]:
    print(scientist["name"], "→", scientist["url"])
```

```python
einstein = next(
    s for s in scientists
    if "Albert Einstein" in s["name"]
)

print(einstein)
```

```python
response = requests.get(einstein["url"])

print("Status code:", response.status_code)

soup = BeautifulSoup(response.text, "html.parser")
```

```python
quotes = soup.find_all("div", class_="quotation")

print("عدد الاقتباسات:", len(quotes))
for i, quote in enumerate(quotes[:5], start=1):
    print(f"\nQuote {i}:")
    print(quote.get_text(" ", strip=True))
```

```python
data = []

for quote in quotes:
    text = quote.get_text(" ", strip=True)

    if text:
        data.append({
            "scientist": einstein["name"],
            "quote": text
        })

print(data[:3])
```

تحويلها إلى DataFrame

```python
import pandas as pd

df = pd.DataFrame(data)

df.head()
```

```python
all_quotes = []

for i, scientist in enumerate(scientists):

    try:
        response = requests.get(
            scientist["url"],
            timeout=10
        )

        if response.status_code != 200:
            print(
                f"{i+1}/{len(scientists)} - {scientist['name']} : ERROR {response.status_code}"
            )
            continue

        soup = BeautifulSoup(response.text, "html.parser")

        quotes = soup.find_all("div", class_="quotation")

        for quote in quotes:

            text = quote.get_text(" ", strip=True)

            if text:
                all_quotes.append({
                    "scientist": scientist["name"],
                    "quote": text
                })

        print(
            f"{i+1}/{len(scientists)} - "
            f"{scientist['name']} → "
            f"{len(quotes)} quotes"
        )

    except Exception as e:

        print(
            f"{i+1}/{len(scientists)} - "
            f"{scientist['name']} → ERROR: {e}"
        )
```

```python
df = pd.DataFrame(all_quotes)

print("عدد الاقتباسات الكلي:", len(df))

df.head(10)
```

```python
df = df.drop_duplicates(subset=["scientist", "quote"])

print("بعد إزالة التكرار:", len(df))
```

```python
df = df[
    df["quote"].notna() &
    (df["quote"].str.strip() != "")
]

print("عدد الاقتباسات:", len(df))
```

```python
df_1000 = df.head(1000).copy()

print("عدد الاقتباسات:", len(df_1000))
```

```python
df_1000 = df.head(1000)
```
